## User-defined functions

```python
def greet(name):
    return "Hello " + name

print(greet("Dharmik"))
```

```python
def add(a, b=10):
    return a + b

print("Positional:", add(5, 6))
print("Default:", add(5))
print("Keyword:", add(b=20, a=5))
```

```python
def total(*numbers):
    return sum(numbers)

print("Variable arguments:", total(10, 20, 30, 40))
```

```python
def show_student(**details):
    for key, value in details.items():
        print(key, ":", value)

show_student(name="Dharmik", age=21, course="Computer Engineering")
```

## math, random and statistics

```python
import math
import random
import statistics

data = [10, 20, 30, 40, 50]
print("Square root:", math.sqrt(25))
print("Power:", math.pow(2, 3))
print("Random number:", random.randint(1, 100))
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode([1, 2, 2, 3, 4]))
```
