## One Sample t-test and Interpretation

**Subject:** Python for Data Science (BE05000231)

**Aim:** One Sample t-test and Interpretation.

```python
import numpy as np
from scipy.stats import ttest_1samp

marks = np.array([72, 75, 78, 70, 80, 76, 74, 79, 82, 77])
assumed_mean = 75
alpha = 0.05

t_stat, p_value = ttest_1samp(marks, assumed_mean)
print("T-statistic:", t_stat)
print("P-value:", p_value)

if p_value < alpha:
    print("Reject H0: Mean is significantly different from 75")
else:
    print("Fail to reject H0: No significant difference from 75")
```
