```python
import numpy as np
from scipy import stats

data = np.array([12, 15, 14, 10, 18, 20, 16, 15, 14, 17])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data, keepdims=True).mode)
print("Variance:", np.var(data, ddof=1))
print("Standard deviation:", np.std(data, ddof=1))
```

## One-sample t-test

```python
t_stat, p_value = stats.ttest_1samp(data, popmean=15)
print("t-statistic:", t_stat)
print("p-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

## Normality test

```python
statistic, p_value = stats.shapiro(data)
print("Shapiro statistic:", statistic)
print("p-value:", p_value)
```
