```python
import pandas as pd

series = pd.Series([80, 75, 90, 85], index=["A", "B", "C", "D"])
print(series)
```

```python
data = {
    "Name": ["Amit", "Bhavya", "Chetan", "Dharmik"],
    "Department": ["CE", "CE", "IT", "CE"],
    "Marks": [80, 75, 90, 85]
}
df = pd.DataFrame(data)
print(df)
```

## Selection and filtering

```python
print(df["Marks"])
print(df.loc[0:2])
print(df[df["Marks"] >= 80])
```

## Grouping and aggregation

```python
print(df.groupby("Department")["Marks"].mean())
print(df.groupby("Department")["Marks"].agg(["mean", "max", "min", "count"]))
```
