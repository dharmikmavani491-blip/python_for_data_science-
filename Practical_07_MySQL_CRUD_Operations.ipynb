```python
import mysql.connector

# Change these values according to your MySQL setup.
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="YOUR_PASSWORD"
)
cursor = conn.cursor()
print("MySQL connected successfully.")
```

## Create database and table

```python
cursor.execute("CREATE DATABASE IF NOT EXISTS pds_practical")
cursor.execute("USE pds_practical")

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(50),
    marks INT
)
""")
conn.commit()
print("Database and table ready.")
```

## Create / Insert

```python
sql = "INSERT INTO students (name, marks) VALUES (%s, %s)"
values = ("Dharmik", 85)
cursor.execute(sql, values)
conn.commit()
print("Record inserted.")
```

## Read

```python
cursor.execute("SELECT * FROM students")
for row in cursor.fetchall():
    print(row)
```

## Update

```python
sql = "UPDATE students SET marks=%s WHERE name=%s"
cursor.execute(sql, (90, "Dharmik"))
conn.commit()
print("Record updated.")
```

## Delete

```python
sql = "DELETE FROM students WHERE name=%s"
cursor.execute(sql, ("Dharmik",))
conn.commit()
print("Record deleted.")
```

```python
cursor.close()
conn.close()
print("MySQL connection closed.")
```
