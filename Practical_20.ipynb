### Dataset
This practical uses `employee.xlsx`, included in this ZIP package. Keep the Excel file in the same folder as this notebook.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("employee.xlsx")

# Line plot
sample = df.head(10)
plt.figure(figsize=(7,4))
plt.plot(sample["EmployeeNumber"], sample["MonthlyIncome"], marker="o")
plt.title("Line Plot - Monthly Income")
plt.xlabel("Employee Number")
plt.ylabel("Monthly Income")
plt.show()

# Bar chart
counts = df["Department"].value_counts()
plt.figure(figsize=(7,4))
plt.bar(counts.index, counts.values)
plt.title("Bar Chart - Employees by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.show()

# Histogram
plt.figure(figsize=(7,4))
plt.hist(df["Age"], bins=10)
plt.title("Histogram - Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Pie chart
plt.figure(figsize=(6,6))
plt.pie(counts.values, labels=counts.index, autopct="%1.1f%%")
plt.title("Pie Chart - Department")
plt.show()

# Scatter plot
plt.figure(figsize=(7,4))
plt.scatter(df["Age"], df["MonthlyIncome"], alpha=0.5)
plt.title("Scatter Plot - Age vs Monthly Income")
plt.xlabel("Age")
plt.ylabel("Monthly Income")
plt.show()
```
