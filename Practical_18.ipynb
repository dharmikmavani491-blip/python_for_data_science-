## Data Cleaning: Missing Values, Duplicates and Outliers

**Subject:** Python for Data Science (BE05000231)

**Aim:** Data Cleaning: Missing Values, Duplicates and Outliers.

```python
import pandas as pd
import numpy as np

# Small dataset for demonstration
df = pd.DataFrame({
    "Name": ["A", "B", "C", "C", "E", "F"],
    "Age": [20, 21, np.nan, np.nan, 22, 100],
    "Marks": [80, 85, 78, 78, np.nan, 82]
})
print("Original data:
", df)

# Missing values
print("Missing values:
", df.isnull().sum())
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Duplicate records
df = df.drop_duplicates()

# Outlier treatment using IQR
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df["Age"] = df["Age"].clip(lower, upper)

print("Cleaned data:
", df)
```
