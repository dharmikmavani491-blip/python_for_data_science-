**Objective:** Calculate and interpret mean, median, mode, variance, standard deviation, range, quartiles and IQR on the supplied employee dataset.

**Note:** Run the cells from top to bottom. This notebook is written in a simple practical/lab style.

## Load the given employee dataset

```python
import pandas as pd
import numpy as np

file_path = "employee.csv"
df = pd.read_csv(file_path)
print("Shape:", df.shape)
df.head()
```

## Select a numerical column

```python
data = df["MonthlyIncome"].dropna()
print("Count:", len(data))
print("Minimum:", data.min())
print("Maximum:", data.max())
```

## Mean, Median and Mode

```python
print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode().tolist())
```

## Variance and Standard Deviation

```python
print("Variance:", data.var())
print("Standard Deviation:", data.std())
```

## Range, Quartiles and IQR

```python
q1 = data.quantile(0.25)
q2 = data.quantile(0.50)
q3 = data.quantile(0.75)
iqr = q3 - q1

print("Range:", data.max() - data.min())
print("Q1:", q1)
print("Q2 / Median:", q2)
print("Q3:", q3)
print("IQR:", iqr)
```

## Summary using describe()

```python
print(data.describe())
```
