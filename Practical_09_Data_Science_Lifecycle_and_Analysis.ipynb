```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = "employee.csv"
df = pd.read_csv(file_path)
print(df.shape)
print(df.head())
print(df.info())
```

## Data checking / cleaning

```python
print("Missing values:")
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())

# Remove duplicate records for analysis if any
analysis_df = df.drop_duplicates().copy()
print("Shape after duplicate removal:", analysis_df.shape)
```

## Univariate Analysis - Age

```python
print(analysis_df["Age"].describe())

plt.figure(figsize=(7, 4))
plt.hist(analysis_df["Age"], bins=15)
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.title("Univariate Analysis - Age")
plt.show()
```

## Bivariate Analysis - Age vs MonthlyIncome

```python
plt.figure(figsize=(7, 4))
plt.scatter(analysis_df["Age"], analysis_df["MonthlyIncome"])
plt.xlabel("Age")
plt.ylabel("Monthly Income")
plt.title("Bivariate Analysis - Age vs Monthly Income")
plt.show()

print(analysis_df[["Age", "MonthlyIncome"]].corr())
```

## Multivariate Analysis

```python
selected = analysis_df[["Age", "MonthlyIncome", "TotalWorkingYears", "YearsAtCompany", "JobLevel"]]
print(selected.corr())

print("\nGrouped analysis by JobLevel:")
print(analysis_df.groupby("JobLevel")[["MonthlyIncome", "Age", "TotalWorkingYears"]].mean())
```
