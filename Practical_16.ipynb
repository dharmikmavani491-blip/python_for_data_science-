## Normal, Exponential Distributions and Central Limit Theorem

**Subject:** Python for Data Science (BE05000231)

**Aim:** Normal, Exponential Distributions and Central Limit Theorem.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, expon

x = np.linspace(-4, 4, 200)
plt.figure(figsize=(7,4))
plt.plot(x, norm.pdf(x, 0, 1))
plt.title("Normal Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()

x = np.linspace(0, 8, 200)
plt.figure(figsize=(7,4))
plt.plot(x, expon.pdf(x, scale=1))
plt.title("Exponential Distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()

# Central Limit Theorem
population = np.random.exponential(scale=10, size=10000)
sample_means = [np.mean(np.random.choice(population, 30)) for _ in range(1000)]
plt.figure(figsize=(7,4))
plt.hist(sample_means, bins=30)
plt.title("Central Limit Theorem - Sampling Means")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()
```
