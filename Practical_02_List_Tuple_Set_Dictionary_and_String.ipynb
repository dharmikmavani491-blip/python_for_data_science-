## List

```python
numbers = [10, 20, 30]
print("Original:", numbers)
numbers.append(40)
numbers.insert(1, 15)
numbers.remove(20)
print("After operations:", numbers)
print("Length:", len(numbers))
print("Sorted:", sorted(numbers))
```

## Tuple

```python
student = ("Dharmik", 21, "Computer Engineering")
print(student)
print("Name:", student[0])
print("Count of 21:", student.count(21))
print("Index of 21:", student.index(21))
```

## Set

```python
numbers = {10, 20, 30, 20}
print("Set:", numbers)
numbers.add(40)
numbers.remove(10)
print("After operations:", numbers)
print("Union:", numbers.union({50, 60}))
print("Intersection:", numbers.intersection({20, 40, 70}))
```

## Dictionary

```python
student = {"name": "Dharmik", "age": 21, "course": "Computer Engineering"}
print(student)
student["age"] = 22
student["city"] = "Surat"
print("Keys:", student.keys())
print("Values:", student.values())
print("Items:", student.items())
```

## String

```python
text = "Python for Data Science"
print(text.upper())
print(text.lower())
print(text.replace("Python", "PDS"))
print("Split:", text.split())
print("Length:", len(text))
print("Starts with Python:", text.startswith("Python"))
```
