## 1. Variables and Data Types

```python
name = "Dharmik"
age = 21
percentage = 82.5
is_student = True

print("Name:", name)
print("Age:", age)
print("Percentage:", percentage)
print("Student:", is_student)
print(type(name))
print(type(age))
print(type(percentage))
print(type(is_student))
```

## 2. Arithmetic Operators

```python
a = 10
b = 3
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Power:", a ** b)
```

## 3. Assignment Operators

```python
x = 10
x += 5
print("x += 5:", x)
x -= 2
print("x -= 2:", x)
x *= 3
print("x *= 3:", x)
x /= 2
print("x /= 2:", x)
```

## 4. Comparison and Logical Operators

```python
a = 10
b = 20
print(a == b)
print(a != b)
print(a < b)
print(a > b)
print(a <= b)
print(a >= b)
print(a < b and b > 15)
print(a > b or b > 15)
print(not(a > b))
```

## 5. Membership and Identity Operators

```python
numbers = [10, 20, 30]
print(20 in numbers)
print(50 not in numbers)

x = numbers
y = numbers.copy()
print("x is numbers:", x is numbers)
print("y is numbers:", y is numbers)
```

## 6. Bitwise Operators

```python
a = 5
b = 3
print("AND:", a & b)
print("OR:", a | b)
print("XOR:", a ^ b)
print("NOT:", ~a)
print("Left Shift:", a << 1)
print("Right Shift:", a >> 1)
```
