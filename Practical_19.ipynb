### Dataset
This practical uses `employee.xlsx`, included in this ZIP package. Keep the Excel file in the same folder as this notebook.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.read_excel("employee.xlsx")

# Slicing
print("Slicing:
", df.loc[0:4, ["Age", "Department", "MonthlyIncome"]])

# Filtering
print("Filtering:
", df[df["MonthlyIncome"] > 10000].head())

# Sorting
print("Sorting:
", df.sort_values("MonthlyIncome", ascending=False)[["Age", "MonthlyIncome"]].head())

# Aggregation
grouped = df.groupby("Department")["MonthlyIncome"].agg(["mean", "sum", "count"])
print("Aggregation:
", grouped)

# Concatenating two dataframes
part1 = df[["EmployeeNumber", "Age"]].head(3)
part2 = df[["EmployeeNumber", "Age"]].tail(3)
combined = pd.concat([part1, part2], ignore_index=True)
print("Concatenated:
", combined)

# Normalization
scaler = MinMaxScaler()
df[["Age_Normalized", "Income_Normalized"]] = scaler.fit_transform(df[["Age", "MonthlyIncome"]])
print(df[["Age", "MonthlyIncome", "Age_Normalized", "Income_Normalized"]].head())
```
