## if / if-else / elif

```python
marks = 78
if marks >= 90:
    grade = "A+"
elif marks >= 75:
    grade = "A"
elif marks >= 60:
    grade = "B"
else:
    grade = "C"
print("Grade:", grade)

if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

## while loop

```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

## for loop

```python
for i in range(1, 6):
    print(i)
```

## Nested loop

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

## break, continue and pass

```python
for i in range(1, 8):
    if i == 4:
        break
    print("break:", i)

for i in range(1, 6):
    if i == 3:
        continue
    print("continue:", i)

for i in range(3):
    if i == 1:
        pass
    print("pass:", i)
```
